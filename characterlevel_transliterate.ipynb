```python
ARABIZI_TO_ARABIC = {
    # Digits
    "2": "ء",
    "3": "ع",
    "5": "خ",
    "6": "ط",
    "7": "ح",
    "8": "ه",
    "9": "ق",

    # Digraphs
    "ch": "ش",
    "sh": "ش",
    "kh": "خ",
    "gh": "غ",
    "dh": "ذ",
    "th": "ث",
    "ou": "و",
    "oo": "و",
    "aa": "ا",
    "ee": "ي",
    "ii": "ي",

    # Latin characters
    "a": "ا",
    "b": "ب",
    "c": "ك",
    "d": "د",
    "e": "",
    "f": "ف",
    "g": "گ",
    "h": "ه",
    "i": "ي",
    "j": "ج",
    "k": "ك",
    "l": "ل",
    "m": "م",
    "n": "ن",
    "o": "و",
    "p": "پ",
    "q": "ق",
    "r": "ر",
    "s": "س",
    "t": "ت",
    "u": "و",
    "v": "ڤ",
    "w": "و",
    "x": "كس",
    "y": "ي",
    "z": "ز"
}
```

```python
def transliterate(text, mapping):
    result = []
    i = 0
    keys = sorted(mapping, key=len, reverse=True)

    while i < len(text):
        match = next(
            (key for key in keys if text.startswith(key, i)),
            None
        )

        if match is not None:
            result.append(mapping[match])
            i += len(match)
        else:
            result.append(text[i])
            i += 1

    return "".join(result)
```

```python
transliterate("khoubz", ARABIZI_TO_ARABIC)
```
