```python
import pandas as pd
```

```python
ds = pd.read_csv('data/raw/doda_train.csv')
ds.head()
```

```python
ds.shape
```

```python
ds.info()
```

```python
clean_ds = ds.dropna()
```

```python
# removing whitespace from the start and end of each text value
clean_ds["Darija_latin"] = clean_ds["Darija_latin"].str.strip()
clean_ds["Darija_arabic"] = clean_ds["Darija_arabic"].str.strip()
```

```python

clean_ds = clean_ds.replace({"Darija_latin": {"": pd.NA},"Darija_arabic": {"": pd.NA}})
```

```python
clean_ds = clean_ds.dropna(subset=["Darija_latin", "Darija_arabic"])
```

```python
clean_ds = clean_ds.drop_duplicates(subset=["Darija_latin", "Darija_arabic"])
```

```python
conflicts = clean_ds.groupby("Darija_latin")["Darija_arabic"].nunique()
conflicts = conflicts[conflicts > 1]
```

```python
clean_ds.to_csv("data/processed/doda_train_clean.csv", index=False)
```
